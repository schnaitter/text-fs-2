```python
import requests
from bs4 import BeautifulSoup, Tag, Comment
```

## Abruf einer Website

```python
# Set url to the URL of the senatskanzlei Berlin
url = "https://www.berlin.de/rbmskzl/"
```

Now we use the 'get' method from the requests library to perform the HTTP call (!link to the markdown chapter 'Website-Exploration I: Einführung in die Methode des Scraping, Scraping einer Website' above, where we mention the HTTP calls) and access the webpage by url:

```python
# Perform get request, save HTML in the variable response
response = requests.get(url)

# Check if the get request was successful
if response.status_code == 200:
    print("Die Abfrage des HTML-Codes war erfolgreich")
else:
    print(f"Die Abfrage war nicht erfolgreich! Der Fehlercode ist: {response.status_code}")
```

Now we can access the entire HTML content by using the '.text' method:

```python
# Get the HTML text from the reponse
html_text = response.text

# display the first 500 characters of the HTML
html_text[:500]
```

## Extraktion von Text aus HTML

To parse HTML we will use the Beatuiful Soup library. 

```python
soup = BeautifulSoup(html_text)
```

Right now we will only use it to go over tags and get their text. We will ignore certain tags which contain text not displayed by browsers (such as javascript code in the `<script>` tag, metadata in the `meta` tag and so on)

```python
# Only get text that is visible on the website 
# Inlcudes ads, pointers to more info and so on 

def tag_visible(element):
    if element.parent.name in ['style', 'script', 'head', 'title', 'meta', '[document]']:
        return False
    if isinstance(element, Comment):
        return False
    return True

def text_from_html(soup):
    texts = soup.findAll(string=True)
    visible_texts = filter(tag_visible, texts)  
    return [t.strip() for t in visible_texts if len(t) > 1]
```

```python
text = text_from_html(soup)
```
